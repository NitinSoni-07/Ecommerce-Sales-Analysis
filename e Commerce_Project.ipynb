```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.colors as colors
pio.templates.default = "plotly_white"

```

```python
data = pd.read_csv("Sample - Superstore.csv", encoding = 'latin-1')
print(data)
```

```python
print(data.head())
```

```python
print(data.describe())
```

```python
print(data.info())
```

```python
data['Order Date'] = pd.to_datetime(data['Order Date'])
data.info()
```

```python
data['Ship Date'] = pd.to_datetime(data['Ship Date'])  # change data types
data.info()
```

```python
data['order_month'] = data['Order Date'].dt.month
data['order_year'] = data['Order Date'].dt.year
data['order_day_of_week'] = data['Order Date'].dt.dayofweek
```

```python
print(data.head())
```

# Monthly sales Analysis

```python
sales_by_month = data.groupby('order_month')['Sales'].sum().reset_index()

sales_by_month
```

```python
fig = px.line(sales_by_month, x='order_month', y='Sales', title='Sales by month')
fig.show()
```

# Sales by Category

```python

sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()

sales_by_category
```

```python
fig = px.line(sales_by_category,
              x= 'Category',
              y = 'Sales',
              )
fig.show()
```

```python
# Group data
sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()

# Create pie chart
fig = px.pie(
    sales_by_category,
    values='Sales',
    names='Category',   # <-- use exact column name
    hole=0.5,
    color_discrete_sequence=["#00BFFF", "#32CD32", "#FFA500"]
)
fig.update_layout(
   plot_bgcolor = "black",
   paper_bgcolor = "black",
   font_color="white"

)

# Show chart
fig.show()
```

# Segment wise sales

```python
sales_profit_by_segment = data.groupby('Segment').agg({
    'Sales': 'sum',
    'Profit': 'sum'
}).reset_index()

color_palette = colors.qualitative.Pastel1

fig = go.Figure()

fig.add_trace(go.Bar(
    x=sales_profit_by_segment['Segment'],
    y=sales_profit_by_segment['Sales'],
    name='Sales',
    marker_color=color_palette[0]
))

fig.add_trace(go.Bar(
    x=sales_profit_by_segment['Segment'],
    y=sales_profit_by_segment['Profit'],
    name='Profit',
    marker_color=color_palette[1]
))

fig.update_layout(
    title='Sales and Profit Analysis by Customer Segment',
    xaxis_title='Customer Segment',
    yaxis_title='Amount'
)

fig.show()
```

# sales to profit ratio

```python
sales_profit_by_segment = data.groupby('Segment').agg({
    'Sales': 'sum',
    'Profit': 'sum'
}).reset_index()

sales_profit_by_segment['sales_to_profit_ratio'] = (
    sales_profit_by_segment['Sales'] /
    sales_profit_by_segment['Profit']
)

print(
    sales_profit_by_segment[
        ['Segment', 'sales_to_profit_ratio']
    ]
)
```
